## Comparación de métodos iterativos para resolver la ecuación de Laplace

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Usando Gauss-Seidel

N = 50
V = np.zeros((N, N))
V[0, :] = 100  # Condición de frontera

tolerance = 1e-4
max_iterations = 10000

for iteration in range(max_iterations):

    Vold = V.copy()

    for i in range(1, N-1):
        for j in range(1, N-1):
            V[i, j] = 0.25 * (V[i+1, j] + V[i-1, j] + V[i, j+1] + V[i, j-1])

    error = np.max(np.abs(V - Vold))

    if iteration % 100 == 0:
        print(f"Iteración {iteration}, error: {error}")

    if error < tolerance:
        print(f"Convergencia alcanzada en la iteración {iteration}")
        break

x = np.arange(0,N,1)
y = np.arange(0,N,1)
X, Y = np.meshgrid(x, y)
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, V, cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('V')
ax.set_title('Potencial eléctrico')
plt.show()
```

```python
# Ahora usando Jacobi

N = 50
V = np.zeros((N, N))
V[0, :] = 100  # Condición de frontera

tolerance = 1e-4
max_iterations = 10000

for iteration in range(max_iterations):

    Vnew = V.copy()

    for i in range(1, N-1):
        for j in range(1, N-1):
            Vnew[i, j] = 0.25 * (V[i+1, j] + V[i-1, j] + V[i, j+1] + V[i, j-1])

    error = np.max(np.abs(Vnew - V))

    if iteration % 100 == 0:
        print(f"Iteración {iteration}, error: {error}")

    if error < tolerance:
        print(f"Convergencia alcanzada en la iteración {iteration}")
        break

    V = Vnew.copy()

x = np.arange(0,N,1)
y = np.arange(0,N,1)
X, Y = np.meshgrid(x, y)
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, V, cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('V')
ax.set_title('Potencial eléctrico')
plt.show()
```

## Capacitor de placas paralelas

$V_{superior} = 100 V$ y $V_{inferior} = -100 V$ La caja externa está aterrizada $V = 0$ El objetivo es resolver numéricamente:

$$\nabla^2 V = 0$$

```python
# Resolviendo problema del capacitor
# Parámetros del problema
N = 100              # nodos por lado
w = 50               # ancho de las placas (en nodos)
d = 20               # separación entre placas (en nodos)
V_top = 100.0       # voltaje placa superior
V_bot = -100.0       # voltaje placa inferior

tolerance = 1e-4
max_iterations = 10000

# Inicialización
V = np.zeros((N, N))

mask = np.zeros((N, N)) # Máscara para las placas

# Definir las placas
V[30,30:70] = V_top
mask[30,30:70] = 1
V[60,30:70] = V_bot
mask[60,30:70] = 1

# Iteración de Gauss-Seidel
for iteration in range(max_iterations):

    Vold = V.copy()

    for i in range(1, N-1):
        for j in range(1, N-1):
            if mask[i, j] == 0:  # Solo actualizar si no es una placa
                V[i, j] = 0.25 * (V[i+1, j] + V[i-1, j] + V[i, j+1] + V[i, j-1])

    error = np.max(np.abs(V - Vold))

    if iteration % 100 == 0:
        print(f"Iteración {iteration}, error: {error}")

    if error < tolerance:
        print(f"Convergencia alcanzada en la iteración {iteration}")
        break

# Visualización
x = np.arange(0,N,1)
y = np.arange(0,N,1)
X, Y = np.meshgrid(x, y)
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, V, cmap='coolwarm', edgecolor='k', linewidth=0.2)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('V')
ax.set_title('Potencial eléctrico de un capacitor realista')
plt.show()
```

```python
# Líneas equipotenciales
plt.figure(figsize=(7,6))
cont = plt.contourf(X, Y, V, levels=40, cmap='coolwarm')
plt.colorbar(label='Voltaje (V)')
plt.contour(X, Y, V, levels=20, colors='black', linewidths=0.5)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Líneas equipotenciales del capacitor')
plt.show()
```

Ahora las placas del capacitor ya no están conectadas a una batería, en su lugar la placa superior tiene una densidad de carga uniforme $+\rho$ y la inferior $-\rho$. El objetivo es resolver numéricamente:
$$\nabla^2 V = -\frac{\rho}{\epsilon_0}$$
En las placas, y:
$$\nabla^2 V = 0$$
En el resto del espacio. Las condiciones de frontera son las mismas que antes.

La ecuación discretizada para el interior de las placas es:
$$V_{i,j} = \frac{1}{4} \left( V_{i+1,j} + V_{i-1,j} + V_{i,j+1} + V_{i,j-1} \right) + \frac{\Delta^2 \rho_{i,j}}{4\epsilon_0}$$
Mientras que para el resto del espacio es la misma que antes:
$$V_{i,j} = \frac{1}{4} \left( V_{i+1,j} + V_{i-1,j} + V_{i,j+1} + V_{i,j-1} \right)$$

La densidad de carga se modela como una matriz.

```python
# Resolviendo problema del capacitor con densidad de carga en las placas
# Placa superior con carga positiva y placa inferior con carga negativa
rho0 = 100.0
epsilon0 = 1.0
rho = np.zeros((N, N))
rho[30,30:70] = +rho0
rho[60,30:70] = -rho0
# Parámetros del problema
N = 100              # nodos por lado
delta = 1.0          # tamaño del paso
w = 50               # ancho de las placas (en nodos)
d = 20               # separación entre placas (en nodos)
V_top = 100.0       # voltaje placa superior
V_bot = -100.0       # voltaje placa inferior

tolerance = 1e-5
max_iterations = 10000

# Inicialización
V = np.zeros((N, N))

mask = np.zeros((N, N)) # Máscara para las placas

# Definir las placas
#V[30,30:70] = V_top
#mask[30,30:70] = 1
#V[60,30:70] = V_bot
#mask[60,30:70] = 1

# Iteración de Gauss-Seidel
for iteration in range(max_iterations):

    Vold = V.copy()

    for i in range(1, N-1):
        for j in range(1, N-1):
            V[i, j] = 0.25 * (V[i+1, j] + V[i-1, j] + V[i, j+1] + V[i, j-1]) + (delta**2 * rho[i, j]) / (4 * epsilon0)

    error = np.max(np.abs(V - Vold))

    if iteration % 100 == 0:
        print(f"Iteración {iteration}, error: {error}")

    if error < tolerance:
        print(f"Convergencia alcanzada en la iteración {iteration}")
        break

# Visualización
x = np.arange(0,N,1)
y = np.arange(0,N,1)
X, Y = np.meshgrid(x, y)
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, V, cmap='coolwarm', edgecolor='k', linewidth=0.2)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('V')
ax.set_title('Potencial eléctrico de un capacitor realista')
plt.show()

# Líneas equipotenciales
plt.figure(figsize=(7,6))
cont = plt.contourf(X, Y, V, levels=40, cmap='coolwarm')
plt.colorbar(label='Voltaje (V)')
plt.contour(X, Y, V, levels=20, colors='black', linewidths=0.5)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Líneas equipotenciales del capacitor')
plt.show()
```

# Problema número 1

Considere una región cuadrada bidimensional discretizada mediante una malla uniforme de tamaño NxN. Utilice N = 50

El potencial eléctrico V(x,y) satisface la ecuación de Poisson:

$$\nabla^2 V(x,y) = -\frac{\rho(x,y)}{\epsilon_0}$$

donde $\rho(x,y)$ es la densidad de carga y $\epsilon_0$ es la permitividad del vacío. Suponga que la densidad de carga está dada por:

$$\rho(x,y) = \rho_0 = 50$$

Suponga que en el borde superior de la caja se tiene un potencial de $V_0 = 500$ mientras que los otros tres lados de la caja están a tierra (V = 0). 

Existe una pequeña región cargada en el centro de la caja, con una densidad de carga $\rho_0$.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Usando Gauss-Seidel

N = 50
delta = 1.0  # Espaciado de la malla
epsilon0 = 1.0 # 8.854e-12  # Permitividad del vacío
rho0 = 50  # Densidad de carga
rho = np.zeros((N, N))
# Región cargada centrada pero no puntual
center = N // 2
radius = 5
for i in range(center - radius, center + radius):
    for j in range(center - radius, center + radius):
        rho[i, j] = rho0


V = np.zeros((N, N))
V[0, :] = 100  # Condición de frontera superior
V[-1, :] = 0   # Condición de frontera inferior
V[:, 0] = 0    # Condición de frontera izquierda
V[:, -1] = 0   # Condición de frontera derecha

tolerance = 1e-4
max_iterations = 10000

for iteration in range(max_iterations):

    Vold = V.copy()

    for i in range(1, N-1):
        for j in range(1, N-1):
            V[i, j] = 0.25 * (V[i+1, j] + V[i-1, j] + V[i, j+1] + V[i, j-1]) + (delta**2 * rho[i, j]) / (4 * epsilon0)

    error = np.max(np.abs(V - Vold))

    if iteration % 100 == 0:
        print(f"Iteración {iteration}, error: {error}")

    if error < tolerance:
        print(f"Convergencia alcanzada en la iteración {iteration}")
        break

x = np.arange(0,N,1)
y = np.arange(0,N,1)
X, Y = np.meshgrid(x, y)
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, V, cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('V')
ax.set_title('Potencial eléctrico')
plt.show()
```

```python
# Líneas equipotenciales
plt.figure(figsize=(7,6))
cont = plt.contourf(X, Y, V, levels=40, cmap='coolwarm')
plt.colorbar(label='Voltaje (V)')
plt.contour(X, Y, V, levels=20, colors='black', linewidths=0.5)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Líneas equipotenciales del capacitor')
plt.show()
```

```python
# Utilizando el potencial obtenido calcular las componentes del campo eléctrico
# Ex = - \partial V / partial x
# Ey = - \partial V / partial y
# Hay que aproximar las derivadas usando diferencias finitas centrales.
# Y por último graficar el campo eléctrico usando quiver.

Ex = np.zeros_like(V)
Ey = np.zeros_like(V)

for i in range(1, N-1):
    for j in range(1, N-1):
        Ex[i, j] = -(V[i+1, j] - V[i-1, j]) / (2 * delta)
        Ey[i, j] = -(V[i, j+1] - V[i, j-1]) / (2 * delta)

# Graficar el campo eléctrico usando quiver
plt.figure(figsize=(7,7))
plt.contour(X,Y,V,20)
plt.quiver(X, Y, Ex, Ey)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Campo eléctrico')
plt.axis('equal')
plt.show()
```

# Problema número 2
Carga puntual q que se coloca dentro de un cilindro conductor de radio a y longitud L. 
Las paredes laterales y las tapas del cilindro están a tierra (V = 0). El objetivo es encontrar el potencial eléctrico dentro del cilindro debido a la carga puntual q.
La carga se ubica en $(r_0, \phi_0, z_0)$

$$\nabla^2 V = -\frac{\rho(\mathbf{r})}{\epsilon_0}$$

El laplaciano en cilindricas es:

$$\nabla^2 V = \frac{1}{r} \frac{\partial}{\partial r} \left( r \frac{\partial V}{\partial r} \right) + \frac{1}{r^2} \frac{\partial^2 V}{\partial \phi^2} + \frac{\partial^2 V}{\partial z^2}$$

Las condiciones de frontera son:

$$V(r=a, \phi, z) = 0, \quad V(r, \phi, z=0) = 0, \quad V(r, \phi, z=L) = 0$$

```python
epsilon0 = 8.854e-12  # Permitividad del vacío
q = 1e-9  # Carga puntual
a = 20.0  # Distancia de la carga al centro de la malla
L = 20.0  # Tamaño de la malla

# Posición de la carga
r0 = 8.0
# phi0 = np.pi / 4
z0 = 10.0

# Parámetros numéricos
Nr = 80 #puntos radiales
Nz = 80 #puntos axiales

dr = a / (Nr - 1)
dz = L / (Nz - 1)

r = np.linspace(0, a, Nr)
z = np.linspace(0, L, Nz)
R, Z = np.meshgrid(r, z)

V = np.zeros((Nz,Nr))

# Densidad de carga
rho_charge = np.zeros((Nz,Nr))

ir = int(r0 / dr)
iz = int(z0 / dz)

rho_charge[iz, ir] = q / (dr * dz)  # Carga puntual distribuida en el nodo más cercano

# Parámetros de iteración
tolerance = 1e-5
max_iterations = 5000

#Método de Gauss-Seidel

for iteration in range(max_iterations):
    
    Vold = V.copy()

    for i in range(1, Nz-1):
        for j in range(1, Nr-1):
            r_j = r[j]
            V[i, j] = (1/(2*(1 + (dr/dz)**2))) * (
                (r_j + dr/2)*V[i, j+1] + (r_j - dr/2)*V[i, j-1] + 
                (dr/dz)**2 * (V[i+1, j] + V[i-1, j]) + 
                (dr**2 * rho_charge[i, j]) / epsilon0
            )

    error = np.max(np.abs(V - Vold))

    if iteration % 100 == 0:
        print(f"Iteración {iteration}, error: {error}")

    if error < tolerance:
        print(f"Convergencia alcanzada en la iteración {iteration}")
        break

# Graficar el potencial
plt.figure(figsize=(8,6))
plt.contourf(R, Z, V, levels=50, cmap='viridis')
plt.colorbar(label='Potencial (V)')
plt.xlabel('r')
plt.ylabel('z')
plt.title('Potencial eléctrico de una carga puntual en coordenadas cilíndricas')
plt.show()
```
