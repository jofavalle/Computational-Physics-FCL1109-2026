# Ecuación de Navier-Stokes

$$\frac{D\mathbf{v}}{Dt} = \nu \nabla^2 \mathbf{v} -\frac{1}{\rho}\nabla P(\rho, T, \mathbf{x})$$

Donde: 

$$\frac{D\mathbf{v}}{Dt} = \frac{\partial \mathbf{v}}{\partial t} + (\mathbf{v} \cdot \nabla) \mathbf{v}$$

Es la derivada hidrodinámica.

Forma explícita (Componentes): 

$$\frac{\partial v_x}{\partial t} + v_x \frac{\partial v_x}{\partial x} + v_y \frac{\partial v_x}{\partial y} + v_z \frac{\partial v_x}{\partial z} = \nu \left( \frac{\partial^2 v_x}{\partial x^2} + \frac{\partial^2 v_x}{\partial y^2} + \frac{\partial^2 v_x}{\partial z^2} \right) - \frac{1}{\rho} \frac{\partial P}{\partial x}$$

$$\frac{\partial v_y}{\partial t} + v_x \frac{\partial v_y}{\partial x} + v_y \frac{\partial v_y}{\partial y} + v_z \frac{\partial v_y}{\partial z} = \nu \left( \frac{\partial^2 v_y}{\partial x^2} + \frac{\partial^2 v_y}{\partial y^2} + \frac{\partial^2 v_y}{\partial z^2} \right) - \frac{1}{\rho} \frac{\partial P}{\partial y}$$

$$\frac{\partial v_z}{\partial t} + v_x \frac{\partial v_z}{\partial x} + v_y \frac{\partial v_z}{\partial y} + v_z \frac{\partial v_z}{\partial z} = \nu \left( \frac{\partial^2 v_z}{\partial x^2} + \frac{\partial^2 v_z}{\partial y^2} + \frac{\partial^2 v_z}{\partial z^2} \right) - \frac{1}{\rho} \frac{\partial P}{\partial z}$$

Suposiciones simplificadoras: 
- La presión es independiente de $\rho$ y $T$
- Flujo estacionario ($\frac{\partial \mathbf{v}}{\partial t} = 0$)

Entonces: 

$$\nabla \cdot \mathbf{v} = 0$$

$$(\mathbf{v} \cdot \nabla) \mathbf{v} = -\frac{1}{\rho} \nabla P + \nu \nabla^2 \mathbf{v}$$

## Ecuación de continuidad en 2D

Para el caso bidimensional se asume:

- $\mathbf{v} = (v_x(x,y,t),\, v_y(x,y,t),\,0)$
- $\dfrac{\partial}{\partial z} = 0$
- $v_z = 0$

Entonces, la continuidad para fluido incompresible queda:

$$
\frac{\partial v_x}{\partial x} + \frac{\partial v_y}{\partial y} = 0
$$

Y las ecuaciones de Navier–Stokes en 2D son:

$$
\frac{\partial v_x}{\partial t}
+ v_x \frac{\partial v_x}{\partial x}
+ v_y \frac{\partial v_x}{\partial y}
=
-\frac{1}{\rho}\frac{\partial P}{\partial x}
+ \nu \left(
\frac{\partial^2 v_x}{\partial x^2}
+ \frac{\partial^2 v_x}{\partial y^2}
\right)
$$

$$
\frac{\partial v_y}{\partial t}
+ v_x \frac{\partial v_y}{\partial x}
+ v_y \frac{\partial v_y}{\partial y}
=
-\frac{1}{\rho}\frac{\partial P}{\partial y}
+ \nu \left(
\frac{\partial^2 v_y}{\partial x^2}
+ \frac{\partial^2 v_y}{\partial y^2}
\right)
$$

## Discretización de las ecuaciones 2D y algoritmo SOR (Sobre-Relajación).
### 1) Malla y notación

Sea una malla uniforme en 2D con nodos $(i,j)$, pasos espaciales $\Delta x,\Delta y$ y paso temporal $\Delta t$:

$$
x_i=i\Delta x,\qquad y_j=j\Delta y,\qquad t^n=n\Delta t
$$

Definimos:

$$
u_{i,j}^n \equiv v_x(x_i,y_j,t^n),\qquad
v_{i,j}^n \equiv v_y(x_i,y_j,t^n),\qquad
p_{i,j}^n \equiv P(x_i,y_j,t^n)
$$

---

### 2) Aproximaciones por diferencias finitas

Derivadas temporales (adelante en tiempo):

$$
\frac{\partial u}{\partial t}\Big|_{i,j}^n \approx \frac{u_{i,j}^{n+1}-u_{i,j}^{n}}{\Delta t},\qquad
\frac{\partial v}{\partial t}\Big|_{i,j}^n \approx \frac{v_{i,j}^{n+1}-v_{i,j}^{n}}{\Delta t}
$$

Derivadas espaciales (centrales):

$$
\frac{\partial \phi}{\partial x}\Big|_{i,j} \approx \frac{\phi_{i+1,j}-\phi_{i-1,j}}{2\Delta x},\qquad
\frac{\partial \phi}{\partial y}\Big|_{i,j} \approx \frac{\phi_{i,j+1}-\phi_{i,j-1}}{2\Delta y}
$$

$$
\nabla^2\phi\Big|_{i,j}\approx
\frac{\phi_{i+1,j}-2\phi_{i,j}+\phi_{i-1,j}}{\Delta x^2}
+
\frac{\phi_{i,j+1}-2\phi_{i,j}+\phi_{i,j-1}}{\Delta y^2}
$$

---

### 3) Esquema fraccionado (proyección)

Primero se calcula una velocidad intermedia $\mathbf{v}^*=(u^*,v^*)$ sin presión:

$$
\frac{u_{i,j}^{*}-u_{i,j}^{n}}{\Delta t}
=
-\left(u\frac{\partial u}{\partial x}+v\frac{\partial u}{\partial y}\right)_{i,j}^{n}
+\nu\nabla^2 u_{i,j}^{n}
$$

$$
\frac{v_{i,j}^{*}-v_{i,j}^{n}}{\Delta t}
=
-\left(u\frac{\partial v}{\partial x}+v\frac{\partial v}{\partial y}\right)_{i,j}^{n}
+\nu\nabla^2 v_{i,j}^{n}
$$

Luego, la presión se obtiene de Poisson:

$$
\nabla^2 p^{n+1}
=
\frac{\rho}{\Delta t}
\left(
\frac{\partial u^*}{\partial x}
+
\frac{\partial v^*}{\partial y}
\right)
$$

Finalmente, se corrige la velocidad:

$$
u_{i,j}^{n+1}=u_{i,j}^{*}-\frac{\Delta t}{\rho}\frac{\partial p^{n+1}}{\partial x}\Big|_{i,j},\qquad
v_{i,j}^{n+1}=v_{i,j}^{*}-\frac{\Delta t}{\rho}\frac{\partial p^{n+1}}{\partial y}\Big|_{i,j}
$$

---

### 4) Discretización de Poisson para presión

Con

$$
b_{i,j}=\frac{\rho}{\Delta t}
\left(
\frac{u_{i+1,j}^{*}-u_{i-1,j}^{*}}{2\Delta x}
+
\frac{v_{i,j+1}^{*}-v_{i,j-1}^{*}}{2\Delta y}
\right)
$$

la ecuación discreta es:

$$
\frac{p_{i+1,j}-2p_{i,j}+p_{i-1,j}}{\Delta x^2}
+
\frac{p_{i,j+1}-2p_{i,j}+p_{i,j-1}}{\Delta y^2}
=
b_{i,j}
$$

Despejando:

$$
p_{i,j}
=
\frac{
\Delta y^2\left(p_{i+1,j}+p_{i-1,j}\right)
+
\Delta x^2\left(p_{i,j+1}+p_{i,j-1}\right)
-
b_{i,j}\Delta x^2\Delta y^2
}{
2\left(\Delta x^2+\Delta y^2\right)
}
$$

---

### 5) Método SOR (Successive Over-Relaxation)

Iteración SOR para resolver Poisson:

$$
p_{i,j}^{(k+1)}
=
(1-\omega)p_{i,j}^{(k)}
+
\omega\,
\frac{
\Delta y^2\left(p_{i+1,j}^{(k)}+p_{i-1,j}^{(k+1)}\right)
+
\Delta x^2\left(p_{i,j+1}^{(k)}+p_{i,j-1}^{(k+1)}\right)
-
b_{i,j}\Delta x^2\Delta y^2
}{
2\left(\Delta x^2+\Delta y^2\right)
}
$$

Forma equivalente con residual local: 

$$
p_{i,j}^{(k+1)}-p_{i,j}^{(k)} = \omega\,\frac{r_{i,j}^{(k)}}{a_{i,j}}
$$

donde $r_{i,j}^{(k)}=b_{i,j}-(A p^{(k)})_{i,j}$ y $a_{i,j}$ es el coeficiente diagonal del operador discreto. Si se define el residual normalizado $\tilde r_{i,j}^{(k)}=r_{i,j}^{(k)}/a_{i,j}$, entonces:

$$
p_{i,j}^{(k+1)}-p_{i,j}^{(k)} = \omega\,\tilde r_{i,j}^{(k)}.
$$

donde $1<\omega<2$ acelera la convergencia (típicamente $\omega\approx1.5\text{ a }1.9$).

Criterio de paro (ejemplo):

$$
\frac{\left\|p^{(k+1)}-p^{(k)}\right\|_2}{\left\|p^{(k+1)}\right\|_2}<\varepsilon
$$

---

### 6) Resumen del algoritmo por paso de tiempo

1. Imponer condiciones de frontera para $u^n,v^n,p^n$.  
2. Calcular $u^*,v^*$ con convección + difusión.  
3. Construir $b_{i,j}$ con $u^*,v^*$.  
4. Resolver Poisson de presión con SOR hasta converger.  
5. Corregir velocidades a $u^{n+1},v^{n+1}$.  
6. Verificar continuidad:

$$
\frac{u_{i+1,j}^{n+1}-u_{i-1,j}^{n+1}}{2\Delta x}
+
\frac{v_{i,j+1}^{n+1}-v_{i,j-1}^{n+1}}{2\Delta y}
\approx 0
$$
