## Analogía con electrostática
En electrostática, la ecuación de Poisson es:
$$\nabla^2 \phi = -\frac{\rho}{\epsilon_0}$$
donde $\phi$ es el potencial eléctrico, $\rho$ es la densidad de carga, y $\epsilon_0$ es la permitividad del vacío.
En dinámica de fluidos, la ecuación de Navier-Stokes con vorticidad se puede escribir como:
$$\nabla^2 \psi = -\omega$$
donde $\psi$ es la función de corriente, y $\omega$ es la vorticidad.

## Forma de vorticidad de la ecuación de Navier-Stokes
Tomando el rotacional de la ecuación de velocidad:
$$\nu \nabla^2 \omega = [(\nabla \times \mathbf{u})\cdot \nabla] \omega$$
